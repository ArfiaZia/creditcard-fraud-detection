```python
import warnings
warnings.filterwarnings('ignore')
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC 
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import StackingClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier
import xgboost # type: ignore
from xgboost import XGBClassifier # type: ignore
from sklearn.model_selection import GridSearchCV
```

```python
data = pd.read_csv("creditcard.csv")
```

```python
X= data.drop('Class',axis=1)
Y= data['Class']
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.33, random_state=42 )

```

```python
model_logistic = LogisticRegression(class_weight='balanced')
scores = cross_val_score(model_logistic, X_train, Y_train, cv=10, scoring='accuracy')
print(scores.mean()) 
model_logistic.fit(X_train, Y_train)
y_predict_log = model_logistic.predict(X_test)
accuracy_score(Y_test, y_predict_log)
```

```python
model_knn = KNeighborsClassifier()
scores = cross_val_score(model_knn, X_train, Y_train, cv=10, scoring='accuracy')
print(scores.mean()) 
model_knn.fit(X_train, Y_train)
y_predict_knn = model_knn.predict(X_test)
accuracy_score(Y_test, y_predict_knn)
```

```python
model_nb = GaussianNB()
scores = cross_val_score(model_nb, X_train, Y_train, cv=10, scoring='accuracy')
print(scores.mean()) 
model_nb.fit(X_train, Y_train)
y_predict_nb = model_nb.predict(X_test)
accuracy_score(Y_test, y_predict_nb)
```

```python
model_tree = DecisionTreeClassifier(class_weight='balanced')
scores = cross_val_score(model_tree, X_train, Y_train, cv=10, scoring='accuracy')
print(scores.mean()) 
model_tree.fit(X_train, Y_train)
y_predict_tree = model_tree.predict(X_test)
accuracy_score(Y_test, y_predict_tree)
```

```python
print(classification_report(Y_test, y_predict_log))
print(classification_report(Y_test, y_predict_knn))
print(classification_report(Y_test, y_predict_nb))
print(classification_report(Y_test, y_predict_tree))
```

by checking the classficiation report we conclude that decision tree works best on non dominent or unseen data as its fraud flags are correct 73% and it also catches actual fraud 74%

```python
param_grid = {
    'max_depth': [3, 5, 10, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'criterion': ['gini', 'entropy'],
    'max_features': [None, 'sqrt', 'log2']
}
grid = GridSearchCV(
    DecisionTreeClassifier(random_state=42), param_grid, cv=5,scoring="f1", n_jobs=-1,
# best_params_ gives the combo that maximizes F1 score
)
grid.fit(X_train, Y_train)
best_model = grid.best_estimator_
print(grid.best_params_)
```

```python
y_pred = best_model.predict(X_test)
print("Accuracy:", accuracy_score(Y_test, y_pred))
print(classification_report(Y_test, y_pred))
train_score = best_model.score(X_train, Y_train)
test_score  = best_model.score(X_test, Y_test)
print("Train:", train_score, " Test:", test_score)
```
